# Get dataset
https://scikit-learn.org/stable/datasets/toy_dataset.html

```python
from sklearn.datasets import load_iris
iris_dict = load_iris()
print(iris_dict.keys())
```

```python
print(f"Features: {iris_dict['feature_names']}")
print(iris_dict['data'][0])

print(f"Target: {iris_dict['target_names']}")
print(iris_dict['target'][0])
```

```python
X = iris_dict['data']
y = iris_dict['target']
```

# Create Train + Test Sets

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
print(f"Train: {X_train[25]} -> {y_train[25]}")
print(f"Test: {X_test[3]} -> {y_test[3]}")
```

# Preprocessing

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler() # subtract the mean and divide by the standard deviation

fit_scaler = scaler.fit(X_train)
X_train_scaled = fit_scaler.transform(X_train)
```

```python
scaler = StandardScaler()

fit_scaler = scaler.fit(X_train)
```

```python
print(f"[Raw] Min: {min(X_train[:, 0])}, Max: {max(X_train[:, 0])}")
print(f"[Scaled] Min: {min(X_train_scaled[:, 0])}, Max: {max(X_train_scaled[:, 0])}")
```

# Basic ML Training + Evaluation

```python
from sklearn.neighbors import KNeighborsClassifier

clf = KNeighborsClassifier()
clf = clf.fit(X_train_scaled, y_train)
```

```python
X_test_scaled = scaler.transform(X_test) # only transform the test data (not fit)
y_pred = clf.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
```

# ML Pipeline

```python
from sklearn.pipeline import Pipeline

# create the pipeline
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier())
])

# fit the pipeline
pipeline.fit(X_train, y_train)

# calculate accuracy
accuracy = pipeline.score(X_test, y_test) # notice that it handles the scaling of the X_test
print(f"Accuracy: {accuracy}")
```
