### Import Libraries

```python
import matplotlib.pyplot as plt
from sklearn import datasets, manifold
import umap
```

### Generate Dataset

```python
X, color = datasets.make_s_curve(n_samples=1200, noise=0.1, random_state=24)
print(f"Shape of X: {X.shape}")
```

### Visualize Original 3D S-Curve

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=color, cmap=plt.cm.viridis)
ax.set_title("Original S-Curve in 3D")
plt.show()
```

### Apply Manifold Learning

#### **E-vector based algorithms:**

```python
# -------- Isomap --------
isomap = manifold.Isomap(n_neighbors=15, n_components=2)
X_isomap = isomap.fit_transform(X)
print(f"Isomap - New shape of X: {X_isomap.shape}")
```

```python
# -------- LLE --------
lle = manifold.LocallyLinearEmbedding(n_neighbors=15, n_components=2, random_state=42)
X_lle = lle.fit_transform(X)
print(f"LLE - New shape of X: {X_lle.shape}")
```

```python
# -------- Diffusion Map --------
diffusion_map = manifold.SpectralEmbedding(n_components=2, random_state=42)
X_diffusion_map = diffusion_map.fit_transform(X)
print(f"Diffusion Map - New shape of X: {X_diffusion_map.shape}")
```

#### **Repulsion based algorithms:**

```python
# -------- TSNE --------
tnse = manifold.TSNE(n_components=2, random_state=42)
X_tsne = tnse.fit_transform(X)
print(f"TSNE - New shape of X: {X_tsne.shape}")
```

```python
# -------- UMAP --------
umap = umap.UMAP(n_neighbors=15, n_components=2, random_state=42)
X_umap = umap.fit_transform(X)
print(f"UMAP - New shape of X: {X_umap.shape}")
```

### Plot 2D Result

#### **E-vector based algorithms:**

```python
# -------- Isomap Plot --------
plt.figure(figsize=(5, 4))
plt.scatter(X_isomap[:, 0], X_isomap[:, 1], c=color, cmap=plt.cm.viridis)
plt.title("2D Projection by Isomap")
plt.axis('tight')
plt.show()
```

```python
## -------- LLE Plot --------
plt.figure(figsize=(5, 4))
plt.scatter(X_lle[:, 0], X_lle[:, 1], c=color, cmap=plt.cm.viridis)
plt.title("2D Projection by LLE")
plt.axis('tight')
plt.show()
```

```python
# -------- Diffusion Map Plot --------
plt.figure(figsize=(5, 4))
plt.scatter(X_diffusion_map[:, 0], X_diffusion_map[:, 1], c=color, cmap=plt.cm.viridis)
plt.title("2D Projection by Diffusion Map")
plt.axis('tight')
plt.show()
```

#### **Repulsion based algorithms:**

```python
# -------- TSNE Plot --------
plt.figure(figsize=(5, 4))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=color, cmap=plt.cm.viridis)
plt.title("2D Projection by TSNE")
plt.axis('tight')
plt.show()
```

```python
# -------- UMAP Plot --------
plt.figure(figsize=(5, 4))
plt.scatter(X_umap[:, 0], X_umap[:, 1], c=color, cmap=plt.cm.viridis)
plt.title("2D Projection by UMAP")
plt.axis('tight')
plt.show()
```
